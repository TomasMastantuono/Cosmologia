```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import zeta
```

```python
mp = 938.27 #MeV
n_b = 6E-10 / 0.022
B_d = 2.23 #MeV

cte = np.log(12*zeta(3)/np.pi**2) + np.log(n_b)
f_T = lambda T: -3*np.log(T)/2 - (B_d/T) + 3*np.log((mp*T)/np.pi)/2
```

```python
Ts = np.linspace(.01, .1, 200) #MeV

fig, ax = plt.subplots(1, 1, figsize = (6, 6))

ax.axhline(cte, linestyle = '--', color = 'gray')
ax.plot(Ts, f_T(Ts), color = 'green')

ax.axvline(.085, linestyle = '-', color = 'red')

ax.set(xlabel = 'T [MeV]', ylabel = 'Funciones')

ax.grid(alpha = .3)

#fig.savefig('P4-1.png')
```

```python
m_n = 939 #MeV
m_p = 938 #MeV
m_H4 = 3727 #MeV

tau = 886 #s
H0 = 2.2E-18 #s-1
Og = 8.24E-5

T_bbn = 0.085 #MeV
T0 = 2.35E-10 #MeV
Tfo = .8 #MeV

t_bbn = 300 #s
t_fo = (T0/Tfo)**2 / 2 / H0 / np.sqrt(Og)

np_nn = np.exp((m_n - m_p)/Tfo)
abundancia_relativa = np_nn**(-1) * np.exp((t_fo - t_bbn)/tau)

print(f'Abundancia relativa de neutrones: {abundancia_relativa}')
print(f'La abundancia de Helio-4 es: {m_H4 / (m_n + (m_p * (abundancia_relativa**(-1))))}')
```
