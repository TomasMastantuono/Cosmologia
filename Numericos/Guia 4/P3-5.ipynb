# Librerias y constantes

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

O_B = .27
NB = 6E-10 * (O_B / .022)
BD = 2.23 #MeV
```

# Intersección

```python
integrando = lambda x: x**2 / (np.exp(x) - 1)
integral = lambda T: quad(integrando, BD/T, np.inf)[0]

T_values = np.linspace(.07, .09, 1_000)
izq = [4*(T**3)*integral(T)/(2*np.pi)**2 for T in T_values]
der = NB * np.pi**2 / 15 * T_values**4

fig, ax = plt.subplots(figsize = (12, 6))
ax.plot(T_values, izq, label = r'$n_{\gamma}(\epsilon > B_D)$',
        color = 'red', linestyle = '--')
ax.plot(T_values, der, label = r'$\eta_b n_{\gamma}$',
        color = 'blue', linestyle = '--')
ax.axvline(.0862, linestyle = '--', label = r'$T_{BBN} \sim 0.08 [MeV]$', color = 'green')

ax.set(xlabel = 'T [MeV]')
ax.xaxis.set_inverted(True)
ax.grid(alpha = .3)
ax.legend()
fig.savefig('P3-5.png', bbox_inches = 'tight')
```
