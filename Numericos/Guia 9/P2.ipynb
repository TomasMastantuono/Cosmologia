# Librerías y constantes

```python
import numpy as np

C = 3E5 #km/s

OMEGA_M = 0.3
OMEGA_R = 8.23E-5
OMEGA_RH2 = 4.15E-5

Z_EQ = OMEGA_M/OMEGA_R - 1
```

# Item $(a)$

```python
keq = 100/C*np.sqrt(OMEGA_M*(1 + Z_EQ) + OMEGA_R*(1 + Z_EQ)**2) # h/Mpc
print('keq = ', keq)

keq2 = 100/C*np.sqrt(2/OMEGA_RH2)
print('Otra forma: keq = ', keq2, 'Ωrh²')
```
