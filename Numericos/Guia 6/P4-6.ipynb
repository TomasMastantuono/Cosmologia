# Importo librerias y constantes

```python
import numpy as np
from scipy.special import zeta

import matplotlib.pyplot as plt

# Constantes
G = 6.71E-45 #MeV-2
H0 = 1.5E-39 #MeV
MP = 938 #MeV
T0 = 2.35E-10 #MeV
GF = 1.16E-11 #MeV-2

OR = 8.24E-5
```

# Item $(b)$

```python
def O_nu(mn):
    cte = 16 * np.pi**2 * G * np.sqrt(OR) / (3 * H0 * GF**2 * mn**2) * T0
    log = np.log(T0**2 * GF**2 * mn**3 / (H0 * np.sqrt(OR)))
    return cte * log

mns = np.linspace(.5E-3, 10E-3, 10_000) #MeV

fig, ax = plt.subplots(figsize = (14, 6))

ax.plot(mns, O_nu(mns), label = r'$\omega_{\nu,0}$')
ax.axhline(1, linestyle = '--', color = 'gray', label = r'$\Omega_{\nu,0} < 1$')

ax.set(xlabel = r'$m_\nu$', ylabel = r'$\Omega_{\nu,0}(m_\nu)$'), ax.legend()
ax.grid(alpha = .3)

plt.show()
```
