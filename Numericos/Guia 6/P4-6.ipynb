# Importo librerias y constantes

```python
import numpy as np
from scipy.special import zeta

import matplotlib.pyplot as plt

# Constantes
G = 6.71E-45 #MeV-2
H0 = 1.5E-39 #MeV
MP = 938 #MeV
T0 = 2.35E-10 #MeV
GF = 1.16E-11 #MeV-2

OR = 8.24E-5
```

# Item $(b)$

```python
f1 = lambda m_nu: np.log( (T0**2 * GF**2 * m_nu**3) / (H0 * np.sqrt(OR) * 2 * np.pi) )
f2 = lambda m_nu: ( (3*H0*GF**2 * m_nu**2) / (16*np.pi**2 * G *np.sqrt(OR) * T0) )**(2/3)

mns = np.linspace(1, 5, 10_000) #MeV

fig, ax = plt.subplots(figsize = (14, 6))

ax.plot(mns, f1(mns), label = 'Lado izquierdo')
ax.plot(mns, f2(mns), label = 'Lado derecho')

ax.set(xlabel = r'$m_\nu$ [MeV]'), ax.legend()
ax.grid(alpha = .3)

fig.savefig('Acotamiento_masa_nu.png', bbox_inches = 'tight'), plt.show()
```
