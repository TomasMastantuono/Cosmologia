# Librerias y constantes

```python
import numpy as np
from scipy.special import zeta


# Constantes
G = 6.71E-45 #MeV-2
H0 = 1.5E-39 #MeV
MP = 938 #MeV
T0 = 2.35E-7 #MeV
```

# Item $(a)$

```python
RHO_C = 3 * H0**2 / 8 / np.pi / G
rho_x = lambda T:  MP * 4 * zeta(3) / 45 * T**3

o_x0 = rho_x(T0) / RHO_C
print(o_x0)
```
