# Librerias y constantes

```python
import numpy as np
from scipy.special import zeta

import matplotlib.pyplot as plt

# Constantes
G = 6.71E-45 #MeV-2
H0 = 1.5E-39 #MeV
MP = 938 #MeV
T0 = 2.35E-10 #MeV

OR = 8.24E-5
```

# Item $(a)$

```python
RHO_C = 3 * H0**2 / 8 / np.pi / G
rho_x = lambda T:  MP * 4 * zeta(3) / 45 * T**3

o_x0 = rho_x(T0) / RHO_C
print(o_x0)
```

# Item $(b)$

```python
omega = lambda sigma: (8*np.pi*G*np.sqrt(OR)*T0 / 3 / H0) * np.log(
    T0**2 * MP * sigma / H0 / np.sqrt(OR)
)**(3/2) / sigma

# Fijo que Ox = .27 debido a que es DM
OX = .27
sigma = np.linspace(1E-14, 1E-13, 100_000)
sigma_int = .4E-13 #MeV
label_sigma = r'$\sigma_{Int} \approx $' + f'{sigma_int} MeV-2'

fig, ax = plt.subplots(figsize = (14, 6))

ax.plot(sigma, omega(sigma), label = r'$\Omega_{x,0}$')
ax.axhline(OX, linestyle = '--', color = 'gray', label = r'Valor esperado: $\Omega_{x,0} = 0.27$')
ax.axvline(sigma_int, linestyle = ':', color = 'orange', label = label_sigma)

ax.set(xlabel = r'$\sigma$', ylabel = r'$\Omega_{x,0}$'), ax.legend()
ax.grid(alpha = .3)

fig.savefig('sigma_int.png', bbox_inches = 'tight')
plt.show()
print('Sigma = ', sigma_int, 'eV')
```
