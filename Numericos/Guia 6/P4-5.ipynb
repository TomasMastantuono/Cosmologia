# Librerias y constantes

```python
import numpy as np
from scipy.special import zeta


# Constantes
G = 6.71E-45 #MeV-2
H0 = 1.5E-39 #MeV
KB = 8.6E-5 #eV/K

T0 = 2.35E-10 #MeV
TNU0 = 2*KB #eV

O_NU = .27
RHO_C = 3*H0**2 / (8 * np.pi * G) #MeV4
```

# Item $(a)$

Pidiendo que $\Omega_\nu \approx 0.27$ (contenido de *materia oscura*), obtengo un valor estimado para la masa de alguno de los neutrinos:
$$m_\nu = \frac{11\pi^2}{6\zeta(3)} \frac{\rho_c \Omega_{\nu, 0}}{T_0^3}$$

```python
m_nu = lambda O_nu0: 11 * np.pi**2 / (6 * zeta(3)) * (RHO_C * O_nu0 / T0**3)

print('m_nu =', m_nu(O_NU) * 1E6, 'eV')
print('Tnu / m_nu = ', TNU0/(m_nu(O_NU) * 1E6))
```
