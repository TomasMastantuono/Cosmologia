```python
import numpy as np

G = 6.71E-45 #MeV-2
MPL = np.sqrt(8*np.pi*G)**(-1) #Mev
```

```python
v14 = lambda r: (3*np.pi**2 / 2 * AS * r)**(1/4) * MPL

rs = [.036, .01, .001]

for _ in rs:
    print(f'Para r={_} -> V^1/4 = {v14(_)}')
```
