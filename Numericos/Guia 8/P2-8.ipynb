# Librerias y constantes

```python
import numpy as np

MeV = 1E6 #eV
Km = 1/(3.09E19) #MpC

G = 6.71E-45 #MeV-2
HBAR = 6.582E-16 #eV*s
C = 3E5 #km/s
ZREC = 1090
T0 = 2.35E-10 #MeV
H0 = 1.5E-39 #MeV
TRH = 4E15 * 1E3 #MeV
DH = 14200 #Mpc

OR = 8.24E-5
OM = 0.27
OL = 0.73

AEQ2 = (1 + 0.3)**(-1)
AREC = (1 + ZREC)**(-1)
```

# Item $(a)$

```python
# Primera pregunta
d_rec = np.sqrt(OM * (1 + ZREC))**(-1) * H0**(-1) / MeV #eV-1
d_rec = d_rec * (HBAR*C) * Km #Mpc
print(d_rec, 'Mpc')

# Segunda pregunta
I1 = 2/(H0 * np.sqrt(OM)) * (np.sqrt(AEQ2) - np.sqrt(AREC)) / MeV #ev-1
I1 = I1 * (HBAR*C) * Km # Mpc

I2 = 1/(H0*np.sqrt(OL)) * (1/AEQ2 - 1) /MeV #eV-1
I2 = I2 * (HBAR*C) * Km # Mpc
n0 = I1 + I2
print(n0, 'Mpc')

# Tercera pregunta
n_ls = AREC / H0 / np.sqrt(OR) / MeV * (HBAR*C) * Km #Mpc

theta_max = 2 * n_ls / (n0 - n_ls) * (180/np.pi)
print(theta_max)

# Cuarta pregunta
R = 6371 #km
print(2/ (theta_max / (180/np.pi))**2)
```

# Item $(b)$

```python
N_min= np.log((np.pi/2 * DH - n_ls) * np.sqrt(8*np.pi*G/3)*TRH/T0)
print(N_min)
```
