# Librerias y constantes

```python
import numpy as np
import matplotlib.pyplot as plt

B = 13.6 #eV
ETA = 6.1E-10
KB = 8.6E-5 #eV/K
```

# Item $(b)$

```python
f = lambda x: x**2 * np.exp(-x) * .2
xs = np.linspace(0, 27, 1_000)
x_int1 = 3.8E-5
x_int2 = 26.94

fig, ax = plt.subplots(figsize = (12, 5))

ax.plot(xs, f(xs), color = 'red')
ax.axhline(ETA, linestyle = '--')
ax.axvline(x_int1, linestyle = '--', 
          color = 'gray', label = f'x = {x_int1}')

ax.axvline(x_int2, linestyle = '--', 
          color = 'green', label = f'x = {x_int2}')

ax.set(xlabel = 'x = B/T'), ax.grid(alpha = .3)
ax.legend()

fig.savefig('Estimacion_densidades.png', bbox_inches = 'tight')
plt.show()

print(f'La primera temperatura es: {B/x_int1}eV')
print(f'La primera temperatura es: {B/x_int2}eV')
```

# Item $(c)$

```python
TR_num = B / x_int2 / KB #K
TR_lit = 3760 #K

print(f'La temperatura numérica es: {TR_num}, la cual tiene un factor de proporcionalidad {TR_num/TR_lit}')
```
